```python
numbers = list(map(int, input("Enter numbers: ").split()))

print("Smallest element:", max(numbers))
```

```python
n=int(input("enter the number of elements"))
a=[]
for i in range (n):
  m = int(input("enter the element"))
  a.append(m)

  print(a)
print("the smallest element is",min(a))
```
